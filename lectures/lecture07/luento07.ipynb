## Mitä on visuaalinen analytiikka?

<p><blockquote>Finding the question is often more important than finding the answer. Exploratory data analysis is an attitude, a flexibility, and a reliance on display, NOT a bundle of techniques, and should be so taught. 
(<a href="http://doi.org/10.1080/00031305.1980.10482706">Tukey, 1980</a>)</blockquote></p>

<p><blockquote>
Visual analytics is the science of analytical reasoning facilitated by interactive visual interfaces (<a href="https://www.hsdl.org/?abstract&did=485291">Thomas &amp; Cook, 2005</a>).</blockquote></p>

<p><blockquote>
Visual analytics "enable analysts of all types to overcome information overload so that they can detect the expected and discover the unexpected from massive, dynamic, conflicting, and incomplete information while rigorously adhering to privacy and security laws and policies" 
(<a href="https://doi.org/10.1109/MCG.2006.5">Thomas &amp; Cook, 2006</a>). 
</blockquote></p>

<p><blockquote>
"To get the most out of such data, however, users must be able to make sense of it: to pursue questions, uncover patterns of interest, and identify (and potentially correct) errors" 
(<a href="https://queue.acm.org/detail.cfm?id=2146416">Heer &amp; Shneiderman, 2012</a>).
</blockquote></p>

# Seitsemän askelta 

Ben Fry ([2008](https://andor.tuni.fi/permalink/358FIN_TAMPO/i4qs1o/alma9910689665905973)) määrittelee seitsemän askeletta datan visualisoinnille:    
    
    
1. Hanki - luo systemaattinen ja luotettava prosessi aineiston hankkimiseksi
2. Pura - luo aineistoon rakenne ja mallinna se piirteisiin ja luokkiin
3. Suodata - poista kaikki paitsi se aineisto josta olet kiinnostunut
4. Louhi - tunnista aineiston rakenteita tiedon louhinnan eri menetelmin
5. Kuvaile - kuvaile aineisto perustyökaluin
6. Jalosta - jalosta aineisto peruskuvauksista selkeämmäksi ja "houkuttelevammaksi"
7. Vuorovaikuta - luo käyttäjälle mekanismeja manipuloida ja kontrolloida aineistoa

Visualisoinnin prosessi ei eroa merkittävästi datatieteen työprosesissa - eikä sen tietysti ole tarpeenkaan erota. Datan visualisoinnin seitsemän askelta ovat itseasiassa eräs sovellus datatieteen prosessista. Tämä tarkoittaa nimenomaan sitä, että datan kuvaaminen mielekkäässä muodossa on keskeistä datatieteen prosessille.

[Yleisiä periaatteita](https://books.google.fi/books?hl=en&lr=&id=RRswXg4pJhcC&oi=fnd&pg=PR3&dq=visualizing+data&ots=798mYDjWIa&sig=vZE8vLaz3CoPZwf0yqRZQlxTd0M&redir_esc=y#v=onepage&q=visualizing%20data&f=false) aineiston visualisoinnille ovat esimerkiksi:

1. Jokaisella projektilla on omat ainutkertaiset rajoitteensa. Visualisoinnin tavoitteena on kuvata niitä piirteitä, jotka ovat keskeisiä sille aineistolle jota pyritään kuvaamaan.
2. Yleensä vähemmän on enemmän. Mikäli yhdessä kuvassa pyritään kuvaamana liian tarkasti sekä yksityiskohtaisesti jotakin, on todennäköistä, että käyttäjä ei huomaa olennaista. Tavoitteena on siis käyttää niin vähän dataa kuin mahdollista kuvaamaan sitä mikä aineistossa on olennaista.
3. Tunne yleisösi: Kenelle teet kuvaajaa? Miksi he käyttävät sitä ja mitä heidän tulisi oppia tuotetusta visualisoinnista?

Erilaisten visualisointityyppien hahmottaminen kokonaisuutena auttaa alkuun suunnittelussa:

1. Aikasarjat, kuten työttömyysaste kymmenen vuoden aikana. Tähän voi sisältyä myös jokin trendi.
2. Lajitellut luokat, kategorinen muuttuja, joka järjestetään jonkin ominaisuuden mukaan. Esimerkiksi myyjien myyntimäärät
3. Osa kokonaisuutta, kategoriset osakokonaisuudet esitetään kokonaisuuden osana. Esimerkkinä viipalekaaviona.
4. Erokuvaaja, kategorisia muuttujia vertaillaan keskenään ja tarkastellaan eroa. Esimerkiksi budjetoitu ja toteutunut.
5. Frekvenssi, jokin muuttujien ilmentymät valittuna tarkastelujaksona. Esimerkiksi histogrammi.
6. Korrelaatio, jonkin kahden muuttujan välinen suhde.

## Käytännön visualisointiesimerkki

Käydään vielä läpi [Holoviews-kirjastolla toteutettu visualisointiesimerkki](https://holoviews.org/getting_started/Introduction.html).

```python
import pandas as pd
import numpy as np
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')
```

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/pyviz/holoviews/master/examples/assets/station_info.csv'
df = pd.read_csv(url,index_col=0,parse_dates=[0])
df.head()
```

Korrelaatio on helppo kuvaaja ymmärtää. Näemme intuitiivisesti että palveluiden määrällä ja asiakkaiden määrällä on jokin suhde.

```python
scatter = hv.Scatter(df, 'services', 'ridership')
scatter
```

Otetaan mukaan histogrammi aseman perustamisajankohdasta. Mitä kuvaaja tuo lisää? Miksi näiden kuvaajien näyttäminen yhdessä olisi mielekästä?

```python
layout = scatter + hv.Histogram(np.histogram(df['opened'], bins=24), kdims=['opened'])
layout
```

Rikastetaan aineisto lisäämällä siihen taksitiedot. Käytännössä lisäämme New Yorkin kaduille tiedon siitä, mihin taksi ovat jättäneet asiakkaat ja katsotaan syntyykö näistä mielekäs kuvaaja.

```python
import requests
url = 'https://raw.githubusercontent.com/pyviz/holoviews/master/examples/assets/hourly_taxi_data.npz'
taxidata = requests.get(url)

# Ensure that folder data exists
import os
if not os.path.exists('data'):
    os.makedirs('data')

with open('data/hourly_taxi_data.npz', 'wb') as f:
    f.write(taxidata.content)
```

```python
taxi_dropoffs = {hour:arr for hour, arr in np.load('data/hourly_taxi_data.npz').items()}
print('Hours: {hours}'.format(hours=', '.join(taxi_dropoffs.keys())))
print('Taxi data contains {num} arrays (one per hour).\nDescription of the first array:\n'.format(num=len(taxi_dropoffs)))
np.info(taxi_dropoffs['0'])
```

Luodaan kuva ja rajataan se maantieteellisesti ja lisätään tähän taksien kyydistäjättötiedot

```python
bounds = (-74.05, 40.70, -73.90, 40.80)
image = hv.Image(taxi_dropoffs['0'], ['lon','lat'], bounds=bounds)
```

Tehdään kerros, jossa laitamme asemat kartalle.

```python
points = hv.Points(df, ['lon','lat'])
```

Nyt voimme yhdistää kuvaajia sekä kerroksia + ja * merkeillä

```python
layout + image*points
```

Nyt voimme lisätä myös toiminnallisuutta ja antaa käyttäjälle mahdollisuuden toimia datan kanssa. HoloMap ottaa sisäänsä dict-muuttujan jossa olemme määritelleet minkä suhteen toiminnallisuutta halutaan saada aikaan.

```python
dictionary = {int(hour):hv.Image(arr, ['lon','lat'], bounds=bounds) 
              for hour, arr in taxi_dropoffs.items()}
imageDyn = hv.HoloMap(dictionary, kdims='Hour')
```

```python
scatter + imageDyn*points
```

Vaihtoehto Bokeh-kirjastolle on Dash. Katso [vertailu Bokehin ja Dashin vahvuuksista ja heikkouksista](https://blog.sicara.com/bokeh-dash-best-dashboard-framework-python-shiny-alternative-c5b576375f7f).

```python
stationMap=[]
stations=df['service_names'].tolist()
lat = df['lat'].tolist()
lon = df['lon'].tolist()
names = df.index.tolist()
for i in range(len(stations)):
    for item in stations[i]:
        if item.isalpha():
            stationMap.append([item,lat[i],lon[i],names[i]])
stationsDf=pd.DataFrame(stationMap,columns=['line','lat','lon','station'])
stationsDf['color'] = 'red'
stationsDf.head()
```

```python
import matplotlib.cm as cm
ds = hv.Dataset(stationsDf)
stations = ds.to(hv.Scatter, 'lon' ,'lat', 'line')
#stations.opts(cmap='fire')# color='color' )
stations.opts(cmap='fire', color='red')
```
